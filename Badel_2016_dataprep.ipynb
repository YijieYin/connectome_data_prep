```python
import pandas as pd
```

```python
badel = pd.read_excel("data/Badel2016/Badel_et_al_2016_raw.xls", index_col=0)
badel
```

```python
# histogram of all values
badel.unstack().hist(bins=100, figsize=(10, 5), grid=False)
```

```python
# normalise within each cell type (column)
# by the max of the absolute value of the cell type
badel_norm = badel.div(badel.abs().max(axis=0), axis=1)
badel_norm
```

```python
# histogram of all values
badel_norm.unstack().hist(bins=100, figsize=(10, 5), grid=False)
```

```python
# transpose and save
badel_norm.T.to_csv("data/Badel2016/Badel2016.csv")
```
