```python
import pandas as pd
```

## read data

```python
odourresponse_s1 = pd.read_csv('data\Hallem_Carlson_2006\Hallem_Carlson_2006_S1.csv')
odourresponse_s1
```

```python
odourdilution = pd.read_csv('data\Hallem_Carlson_2006\Hallem_Carlson_2006_S2.csv')
odourdilution
```

```python
odourtime = pd.read_csv('data\Hallem_Carlson_2006\Hallem_Carlson_2006_S3.csv')
odourtime['odour'] = odourtime.odour.str.replace("Responses to ", "")
odourtime
```

## S1

```python
# borrow the receptor to glomerulus mapping from DoOR 
odour_mappings = pd.read_csv('data\DoOR\door_receptor_mappings.csv', index_col=0)
receptor_to_glomerulus = dict(zip(odour_mappings['receptor'], odour_mappings['glomerulus']))
odour_mappings
```

```python
# anything present in the Hallem & Carlson data that is not in DoOR?
set('Or'+odourresponse_s1.columns[1:]) - set(odour_mappings.receptor[odour_mappings.adult == True])
```

```python
# map to glomerulus 
odourresponse_s1.columns = [odourresponse_s1.columns[0]] + list(
    "Or" + odourresponse_s1.columns[1:]
)
odourresponse_s1 = odourresponse_s1.rename(columns=receptor_to_glomerulus)
odourresponse_s1.set_index('odour', inplace=True)
odourresponse_s1 = odourresponse_s1.T
odourresponse_s1
```

```python
odourresponse_s1.abs().max(axis=1)
```

We can see that the max firing rate of each glomerulus differs quite a lot. Though there were 100 odours, it might be that none were great for some glomeruli. Assuming ORNs types are more similar than different, maybe should normalise by max firing rate overall. 

```python
# which supplement has the max firing rate? 
odourresponse_s1.abs().max().max(), odourdilution.drop(columns=['odour','dilution']).abs().max().max(), odourtime[['0.0s','0.5s','1.0s','1.5s']].abs().max().max()
```

```python
max_activation = max(
    odourresponse_s1.abs().max().max(),
    odourdilution.drop(columns=["odour", "dilution"]).abs().max().max(),
    odourtime[["0.0s", "0.5s", "1.0s", "1.5s"]].abs().max().max(),
)
```

```python
# normalise: divide by the maximum absolute response
odourresponse_s1 = odourresponse_s1.div(max_activation)
# save
odourresponse_s1.to_csv('data/Hallem_Carlson_2006/odour_response_s1_tidy.csv')
```

## S2

```python
# anything present in the Hallem & Carlson data that is not in DoOR?
set("Or" + odourdilution.columns[2:]) - set(
    odour_mappings.receptor[odour_mappings.adult == True]
)
```

```python
odourdilution.odour.value_counts()
```

```python
odourdilution.odour[~odourdilution.odour.str.contains('pure')].nunique()
```

```python
odourdilution["odour_dilution"] = (
    odourdilution["odour"] + " at " + odourdilution["dilution"].astype(str)
)
odourdilution.drop(columns=["odour", "dilution"], inplace=True)
odourdilution
```

```python
# map to glomerulus
odourdilution.columns = list(
    "Or" + odourdilution.columns[:-1]
) + ["odour_dilution"]
odourdilution = odourdilution.rename(columns=receptor_to_glomerulus)
odourdilution.set_index("odour_dilution", inplace=True)
odourdilution = odourdilution.T
odourdilution
```

```python
# normalise: divide by the maximum absolute response
odourdilution = odourdilution.div(max_activation)
# save
odourdilution.to_csv("data/Hallem_Carlson_2006/odour_dilution_response_s2_tidy.csv")
```

## S3

```python
odourtime['glomerulus'] = odourtime.receptor.map(receptor_to_glomerulus)
odourtime[['0.0s', '0.5s', '1.0s', '1.5s']] = odourtime[['0.0s', '0.5s', '1.0s', '1.5s']]/max_activation
odourtime
```
