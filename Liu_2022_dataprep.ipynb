```python
exdata = pd.read_csv('data/Liu2022/Liu2022_tableS3.csv')
neuronmeta = pd.read_csv('data/Liu2022/Liu2022_tableS1.csv')
exdata
```

```python
# modify cell types for flywire types
neuronmeta['flywire_type'] = neuronmeta['hemibrain type'].replace(
    {
        "LHAV6a3_a": "LHAV6a3",
        "LHAV6a3_b": "LHAV6a3",
        "LHAV6a1_b": "LHAV6a1",
        "LHPV5c1_d": "LHPV5c1",
        "LHAD1b2_c": "LHAD1b2_a,LHAD1b2_c",
        "LHAD1b2_a": "LHAD1b2_a,LHAD1b2_c",
        "LHAV2b2_b": "LHAV2b2b",
        "LHAV2b2_a": "LHAV2b2a",
    },
    
)
# LHAV6a2, LHPD3a1, LHAD1b2_b, LHPD2b1_a, LHPD2b1_b, LHPD2a1, LHAV1a2_b, LHAV1a2_a don't exist in FAFB
jeannetype_to_hemibrain_type = dict(
    zip(
        neuronmeta["Cell type name (Jeanne 2018 nomenclature for LHNs)"],
        neuronmeta["hemibrain type"],
    )
)
jeannetype_to_fafb_type = dict(
    zip(
        neuronmeta["Cell type name (Jeanne 2018 nomenclature for LHNs)"],
        neuronmeta["flywire_type"],
    )
)
```

```python
exdata.loc[:, ["pre"]] = exdata.Connection.str.split("/").str[0]
exdata.loc[:, ["post"]] = exdata.Connection.str.split("/").str[1]
exdata.post = exdata.post.str.replace("'", '')
exdata
```

```python
exdata.loc[:, ['pre_type']] = exdata.pre + '_adPN'
exdata.loc[exdata.pre.isin(["DM1", "VC1"]), 'pre_type'] = exdata[exdata.pre.isin(["DM1", "VC1"])].pre + '_lPN'
exdata
```

```python
exdata.loc[:, ["post_hemibrain_type"]] = exdata.post.map(jeannetype_to_hemibrain_type)
exdata.loc[:, ["post_fafb_type"]] = exdata.post.map(jeannetype_to_fafb_type)
exdata
```

```python
exdata = exdata.loc[
    exdata["mean EPSP amplitude (mV)"].notna(),
    ["pre_type", "post_hemibrain_type", 'post_fafb_type', "mean EPSP amplitude (mV)"],
]
exdata.rename(columns={"mean EPSP amplitude (mV)": "mean_EPSP_amplitude"}, inplace=True)
exdata.to_csv('data/Liu2022/data_tidy.csv', index=False)
```
